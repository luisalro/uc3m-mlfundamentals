# EDA

```python
from scipy import stats

feature_cols_eda = [c for c in train.columns if c.startswith('feat_')]
cat_features_eda = ['feat_16', 'feat_18', 'feat_19']
num_features_eda = [c for c in feature_cols_eda if c not in cat_features_eda]

# Shape & dtypes
print("=" * 60)
print(f"Train shape : {train.shape}")
print(f"Test  shape : {test.shape}")
print("\nData types:")
print(train.dtypes)

#Missing values 
print("\n" + "=" * 60)
print("Missing values (train):")
missing = train.isnull().sum()
missing_pct = (missing / len(train) * 100).round(2)
missing_df = pd.DataFrame({'count': missing, '%': missing_pct})
print(missing_df[missing_df['count'] > 0].sort_values('%', ascending=False))

#Target distribution
print("\n" + "=" * 60)
print("Target (y_train) summary:")
print(train['target'].describe().round(2))
sk = stats.skew(train['target'].dropna())
ku = stats.kurtosis(train['target'].dropna())
print(f"\n  Skewness : {sk:.4f}  {'High positive skew' if sk > 1 else ('High negative skew' if sk < -1 else 'OK')}")
print(f"  Kurtosis : {ku:.4f}")
_, p_sw = stats.shapiro(train['target'].sample(min(5000, len(train)), random_state=42))
print(f"  Shapiro-Wilk p-value : {p_sw:.4e}  → {'NOT normal' if p_sw < 0.05 else 'Normal'}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(train['target'], bins=50, color='steelblue', edgecolor='white')
axes[0].set_title('Target distribution')
axes[0].set_xlabel('target')
stats.probplot(train['target'], dist="norm", plot=axes[1])
axes[1].set_title('Q-Q Plot (target vs Normal)')
plt.tight_layout()
plt.show()

# Numeric features: skewness table 
print("\n" + "=" * 60)
print("Numeric features — Skewness & Kurtosis:")
skew_data = []
for col in num_features_eda:
    s = stats.skew(train[col].dropna())
    k = stats.kurtosis(train[col].dropna())
    skew_data.append({'feature': col, 'skewness': round(s, 3), 'kurtosis': round(k, 3),
                      'flag': 'Check' if abs(s) > 1 else ''})
skew_df = pd.DataFrame(skew_data).sort_values('skewness', ascending=False)
print(skew_df.to_string(index=False))

# Numeric features — distribution plots 
n_cols = 4
n_rows = -(-len(num_features_eda) // n_cols)   # ceiling division
fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, n_rows * 3))
axes = axes.flatten()
for i, col in enumerate(num_features_eda):
    axes[i].hist(train[col].dropna(), bins=30, color='steelblue', edgecolor='white')
    sk_val = stats.skew(train[col].dropna())
    axes[i].set_title(f'{col}\nskew={sk_val:.2f}', fontsize=9)
for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)
plt.suptitle('Numeric Feature Distributions (train)', y=1.01, fontsize=12)
plt.tight_layout()
plt.show()

#  6. Categorical features 
print("\n" + "=" * 60)
print("Categorical features — value counts:")
fig, axes = plt.subplots(1, len(cat_features_eda), figsize=(14, 4))
for ax, col in zip(axes, cat_features_eda):
    vc = train[col].value_counts(dropna=False)
    print(f"\n{col}:\n{vc}")
    ax.bar(vc.index.astype(str), vc.values, color='coral', edgecolor='white')
    ax.set_title(col)
    ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()

#Correlation heatmap 
print("\n" + "=" * 60)
print("Correlation with target (top 10):")
corr_with_target = train[num_features_eda + ['target']].corr()['target'].drop('target').abs().sort_values(ascending=False)
print(corr_with_target.head(10).round(3))

fig, ax = plt.subplots(figsize=(14, 10))
corr_matrix = train[num_features_eda + ['target']].corr()
im = ax.imshow(corr_matrix, cmap='RdBu_r', vmin=-1, vmax=1)
ax.set_xticks(range(len(corr_matrix.columns)))
ax.set_yticks(range(len(corr_matrix.columns)))
ax.set_xticklabels(corr_matrix.columns, rotation=90, fontsize=8)
ax.set_yticklabels(corr_matrix.columns, fontsize=8)
plt.colorbar(im, ax=ax)
ax.set_title('Correlation Heatmap (numeric features + target)')
plt.tight_layout()
plt.show()
```

## 📋 Interpretación del EDA

### 1. Valores Nulos
Todas las **features numéricas** presentan aproximadamente un **9–11% de valores faltantes**, con una distribución muy uniforme entre ellas. Esto sugiere que la ausencia de datos **no es aleatoria para una sola variable**, sino que probablemente responde a un patrón sistemático del proceso de recogida de datos. Las variables categóricas (`feat_16`, `feat_18`, `feat_19`) **no tienen nulos**.

> ✅ Estrategia adoptada: imputación por mediana (SimpleImputer) o por vecinos cercanos (KNNImputer).

---

### 2. Variable Objetivo (`target`)
| Métrica | Valor |
|---|---|
| Media | −8.29 |
| Mediana | −219.6 |
| Rango | [−704.6, +2699.4] |
| **Skewness** | **1.18 ⚠** (sesgo positivo alto) |
| Kurtosis | 0.91 |
| Shapiro-Wilk p | ≈ 0.000 → **NO normal** |

La diferencia notable entre media y mediana, junto con la cola derecha larga visible en el histograma y el Q-Q plot, confirman una **distribución asimétrica positiva**. Esto puede perjudicar modelos lineales y justifica aplicar una transformación del target (`PowerTransformer` con método Yeo-Johnson, que admite valores negativos).

---

### 3. Skewness de las Features Numéricas
| Feature | Skewness | Interpretación |
|---|---|---|
| `feat_1` | **12.42 ⚠** | Extremadamente sesgada — mayoría de valores cerca de 0 con outliers muy altos |
| `feat_14` | **9.73 ⚠** | Igual que `feat_1` — distribución tipo long-tail |
| `feat_6` | **1.51 ⚠** | Sesgo moderado-alto |
| resto | < 1 | Distribuciones aproximadamente simétricas ✅ |

`feat_1` y `feat_14` son candidatas a transformación logarítmica si se usaran modelos lineales. Para modelos de árbol (RF, ExtraTrees) el sesgo **no afecta directamente** al rendimiento.

---

### 4. Correlación con el Target
Las features más correlacionadas linealmente con `target` son:

1. **`feat_11`** → |r| = 0.54 (correlación moderada-alta)

Varias features tienen correlación muy baja o nula con el target (`feat_5`, `feat_4`, `feat_3`), aunque pueden tener relaciones no lineales que el Random Forest sí capturará.

---

### 5. Variables Categóricas
| Variable | Nº clases | Observación |
|---|---|---|
| `feat_16` | 4 (A1–A4) | **Balanceada** (~25% cada clase) ✅ |
| `feat_18` | 2 (C1, No) | **Desbalanceada** — 96.6% es `C1` ⚠ |

`feat_18` y `feat_19` tienen muy poca varianza; su aportación al modelo puede ser limitada. Se recomienda vigilar si el one-hot encoding de estas variables aporta o no señal.

---

### 6. Conclusiones y Decisiones de Modelado
- ✅ **Imputación necesaria** para todas las numéricas (~10% nulos)
- ⚠ **Target con sesgo** → considerar `TransformedTargetRegressor` con `PowerTransformer(yeo-johnson)`
- ⚠ **`feat_1` y `feat_14` muy sesgadas** → irrelevante para árboles, pero importante si se prueban modelos lineales
- ✅ **`feat_11`, `feat_7`, `feat_15`** son las features más informativas según correlación lineal
- ⚠ **`feat_18` y `feat_19`** tienen baja varianza — candidatas a eliminación en feature selection


# Fit

```python

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, PowerTransformer
from sklearn.impute import SimpleImputer, KNNImputer

# ── Feature groups ────────────────────────────────────────────────────────────
feature_cols = [c for c in train.columns if c.startswith('feat_')]
cat_features = ['feat_16', 'feat_18', 'feat_19']

# EDA: feat_1 (skew=12.42) y feat_14 (skew=9.73) extremadamente sesgadas
#      feat_6  (skew=1.51) sesgo alto → aplicar PowerTransformer (Yeo-Johnson)
skewed_features = ['feat_1', 'feat_14', 'feat_6']
num_features    = [c for c in feature_cols if c not in cat_features]
normal_features = [c for c in num_features if c not in skewed_features]

X_train = train[feature_cols]
y_train = train['target']
X_test  = test[feature_cols]

# ── Numeric transformers ──────────────────────────────────────────────────────
# EDA: ~10% nulos en todas las numéricas → KNNImputer más preciso que mediana
# Features normales: KNN imputer + StandardScaler
normal_transformer = Pipeline([
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler',  StandardScaler())
])

# Features sesgadas: KNN imputer + PowerTransformer (Yeo-Johnson admite negativos)
skewed_transformer = Pipeline([
    ('imputer',    KNNImputer(n_neighbors=5)),
    ('power',      PowerTransformer(method='yeo-johnson', standardize=True))
])

# ── Categorical transformer ───────────────────────────────────────────────────
# EDA: feat_18 y feat_19 con baja varianza (~95% en una clase) → se mantienen
#      pero con handle_unknown='ignore' por seguridad
cat_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# ── Main preprocessor ─────────────────────────────────────────────────────────
preprocessor = ColumnTransformer([
    ('normal',  normal_transformer, normal_features),
    ('skewed',  skewed_transformer, skewed_features),
    ('cat',     cat_transformer,    cat_features)
])

print(f"Normal features  ({len(normal_features)}): {normal_features}")
print(f"Skewed features  ({len(skewed_features)}): {skewed_features}")
print(f"Cat    features  ({len(cat_features)}): {cat_features}")
print(f"\nX_train: {X_train.shape}  |  X_test: {X_test.shape}")

results = {}
```

# Evaluation

```python

from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.linear_model import Ridge, Lasso
from sklearn.compose import TransformedTargetRegressor
from sklearn.preprocessing import PowerTransformer
from sklearn.model_selection import cross_val_score
import warnings
warnings.filterwarnings('ignore')

# El nuevo preprocessor ya incorpora:
#   - KNNImputer para todos los nulos (~10%)
#   - PowerTransformer(yeo-johnson) para feat_1, feat_14, feat_6 (skew alto)
# Adicionalmente se envuelve el modelo con TransformedTargetRegressor
# para corregir el sesgo del target (skewness=1.18, distribución no normal)

candidates = {
    'Ridge':         Ridge(),
    'Lasso':         Lasso(max_iter=5000),
    'RandomForest':  RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1),
    'ExtraTrees':    ExtraTreesRegressor(n_estimators=200, random_state=42, n_jobs=-1),
}

target_transformer = PowerTransformer(method='yeo-johnson')

for name, reg in candidates.items():
    base_pipe = Pipeline([('preprocessor', preprocessor), ('regressor', reg)])
    # Modelos lineales se benefician también de la transformación del target
    model = TransformedTargetRegressor(regressor=base_pipe, transformer=target_transformer)
    scores = cross_val_score(model, X_train, y_train, cv=5, scoring='r2', n_jobs=-1)
    results[name] = scores
    print(f"{name:20s}  mean R²={scores.mean():.4f}  std={scores.std():.4f}")
```

# Hyperparameter Tuning - RS CV 

```python

from sklearn.model_selection import RandomizedSearchCV

# Pick the best model from cross-validation results
best_model_name = max(results, key=lambda k: results[k].mean())
best_model = candidates[best_model_name]
print(f"Best model from CV: {best_model_name} (mean R²={results[best_model_name].mean():.4f})")

# Con TransformedTargetRegressor la estructura del pipeline es:
#   TransformedTargetRegressor
#     └── regressor = Pipeline
#                       ├── preprocessor
#                       └── regressor  ← los params se acceden como regressor__regressor__<param>
param_grids = {
    'Ridge': {
        'regressor__regressor__alpha': [0.01, 0.1, 1.0, 10.0, 100.0]
    },
    'Lasso': {
        'regressor__regressor__alpha': [0.001, 0.01, 0.1, 1.0, 10.0]
    },
    'RandomForest': {
        'regressor__regressor__n_estimators':     [100, 200, 400],
        'regressor__regressor__max_depth':        [None, 10, 20, 30],
        'regressor__regressor__min_samples_leaf': [1, 2, 5, 10],
        'regressor__regressor__max_features':     ['sqrt', 'log2', 0.5],
    },
    'ExtraTrees': {
        'regressor__regressor__n_estimators':     [100, 200, 400],
        'regressor__regressor__max_depth':        [None, 10, 20, 30],
        'regressor__regressor__min_samples_leaf': [1, 2, 5, 10],
        'regressor__regressor__max_features':     ['sqrt', 'log2', 0.5],
    },
}

# Reconstruir como TransformedTargetRegressor igual que en Evaluation
best_base_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor',    best_model)
])
best_tt_model = TransformedTargetRegressor(
    regressor=best_base_pipe,
    transformer=PowerTransformer(method='yeo-johnson')
)

search = RandomizedSearchCV(
    best_tt_model,
    param_distributions=param_grids[best_model_name],
    n_iter=30,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

search.fit(X_train, y_train)
print(f"\nBest CV R²: {search.best_score_:.4f}")
print(f"Best params: {search.best_params_}")
```

```python
# Generate predictions with the best tuned model
preds_best = search.best_estimator_.predict(X_test)

output_best = pd.DataFrame({
    'ID':     test['ID'],
    'target': preds_best
})

output_best_path = '../submissions/predictions_best.csv'
output_best.to_csv(output_best_path, index=False)

print(f"Predictions saved to: {output_best_path}")
output_best.head(10)
```

# Improving 1 - Feature Importance & ExtraTrees

```python
from sklearn.ensemble import ExtraTreesRegressor

# Feature importance from the best RF model
# TransformedTargetRegressor stores the FITTED inner pipeline in .regressor_ (with underscore)
# .regressor (no underscore) is the unfitted template — always use .regressor_ after .fit()
best_rf_pipe = search.best_estimator_.regressor_  # fitted Pipeline inside TTR
rf_model = best_rf_pipe.named_steps['regressor']

# Recover feature names after preprocessing
ohe_feature_names = best_rf_pipe.named_steps['preprocessor'] \
    .named_transformers_['cat'] \
    .named_steps['encoder'] \
    .get_feature_names_out(cat_features).tolist()
all_feature_names = num_features + ohe_feature_names

importances = rf_model.feature_importances_
sorted_idx = importances.argsort()[::-1]

plt.figure(figsize=(12, 5))
plt.bar(range(len(importances)), importances[sorted_idx])
plt.xticks(range(len(importances)), [all_feature_names[i] for i in sorted_idx], rotation=90)
plt.title('Feature Importances - Random Forest')
plt.tight_layout()
plt.show()

#  Compare RF vs ExtraTrees 
et_candidates = {
    'RandomForest': RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1),
    'ExtraTrees':   ExtraTreesRegressor(n_estimators=200, random_state=42, n_jobs=-1),
}

for name, reg in et_candidates.items():
    pipe = Pipeline([('preprocessor', preprocessor), ('regressor', reg)])
    scores = cross_val_score(pipe, X_train, y_train, cv=5, scoring='r2', n_jobs=-1)
    results[name] = scores
    print(f"{name:30s}  mean R²={scores.mean():.4f}  std={scores.std():.4f}")
```

## Hyperparameter Tuning - RandomForest / ExtraTrees

```python
# Pick best between RandomForest and ExtraTrees
best_model_name_v2 = max(['RandomForest', 'ExtraTrees'], key=lambda k: results[k].mean())
print(f"Best tree model: {best_model_name_v2} (mean R²={results[best_model_name_v2].mean():.4f})")

param_grids_trees = {
    'RandomForest': {
        'regressor__n_estimators':      [200, 400, 600, 800],
        'regressor__max_depth':         [None, 10, 20, 30, 40],
        'regressor__min_samples_leaf':  [1, 2, 4, 8],
        'regressor__min_samples_split': [2, 5, 10],
        'regressor__max_features':      ['sqrt', 'log2', 0.3, 0.5, 0.7],
        'regressor__bootstrap':         [True, False],
    },
    'ExtraTrees': {
        'regressor__n_estimators':      [200, 400, 600, 800],
        'regressor__max_depth':         [None, 10, 20, 30, 40],
        'regressor__min_samples_leaf':  [1, 2, 4, 8],
        'regressor__min_samples_split': [2, 5, 10],
        'regressor__max_features':      ['sqrt', 'log2', 0.3, 0.5, 0.7],
        'regressor__bootstrap':         [True, False],
    }
}

tree_model_map = {
    'RandomForest': RandomForestRegressor(random_state=42, n_jobs=-1),
    'ExtraTrees':   ExtraTreesRegressor(random_state=42, n_jobs=-1)
}

best_pipe_v2 = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor',    tree_model_map[best_model_name_v2])
])

search_v2 = RandomizedSearchCV(
    best_pipe_v2,
    param_distributions=param_grids_trees[best_model_name_v2],
    n_iter=50,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

search_v2.fit(X_train, y_train)
print(f"\nBest CV R²: {search_v2.best_score_:.4f}")
print(f"Best params: {search_v2.best_params_}")
```

## RF Ensemble - Averaging RF + ExtraTrees

```python
from sklearn.ensemble import VotingRegressor

# Combine RF and ExtraTrees with averaging
pipe_rf = Pipeline([('preprocessor', preprocessor), ('regressor', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))])
pipe_et = Pipeline([('preprocessor', preprocessor), ('regressor', ExtraTreesRegressor(n_estimators=400, random_state=42, n_jobs=-1))])

voting = VotingRegressor([
    ('rf', pipe_rf),
    ('et', pipe_et),
], n_jobs=-1)

scores_voting = cross_val_score(voting, X_train, y_train, cv=5, scoring='r2', n_jobs=-1)
print(f"RF+ExtraTrees Voting  mean R²={scores_voting.mean():.4f}  std={scores_voting.std():.4f}")

# Train on full train set
voting.fit(X_train, y_train)
```

```python
# Generate final predictions - choose best between tuned model and voting ensemble
print(f"Tuned {best_model_name_v2} CV R²:  {search_v2.best_score_:.4f}")
print(f"RF+ExtraTrees Voting CV R²:   {scores_voting.mean():.4f}")

if scores_voting.mean() > search_v2.best_score_:
    final_model = voting
    print("\n→ Using RF+ExtraTrees Voting for final predictions")
else:
    final_model = search_v2.best_estimator_
    print(f"\n→ Using Tuned {best_model_name_v2} for final predictions")

preds_final = final_model.predict(X_test)

output_final = pd.DataFrame({
    'ID':     test['ID'],
    'target': preds_final
})

output_final_path = '../submissions/predictions_final.csv'
output_final.to_csv(output_final_path, index=False)
print(f"\nPredictions saved to: {output_final_path}")
output_final.head(10)
```

# Improving 2 - KNN Imputer

```python
from sklearn.impute import KNNImputer

# KNN Imputer: fills missing values using the k nearest neighbors
# More accurate than median for structured tabular data
num_transformer_knn = Pipeline([
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler',  StandardScaler())
])

preprocessor_knn = ColumnTransformer([
    ('num', num_transformer_knn, num_features),
    ('cat', cat_transformer, cat_features)
])

knn_candidates = {
    'RF + KNN Imputer':  RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1),
    'ET + KNN Imputer':  ExtraTreesRegressor(n_estimators=400, random_state=42, n_jobs=-1),
}

for name, reg in knn_candidates.items():
    pipe = Pipeline([('preprocessor', preprocessor_knn), ('regressor', reg)])
    scores = cross_val_score(pipe, X_train, y_train, cv=5, scoring='r2', n_jobs=-1)
    results[name] = scores
    print(f"{name:35s}  mean R²={scores.mean():.4f}  std={scores.std():.4f}")
```

# Remove low-importance features

```python
from sklearn.feature_selection import SelectFromModel

# Use importances from the best RF to filter out low-signal numeric features
# Features below the mean importance are dropped
importance_threshold = 'mean'  # can also try 'median' or a float like 0.02

# Build selector based on already-fitted RF
selector = SelectFromModel(rf_model, threshold=importance_threshold, prefit=True)
selected_num_mask = selector.get_support()
selected_num_features = [f for f, keep in zip(num_features, selected_num_mask[:len(num_features)]) if keep]
print(f"Numeric features kept: {len(selected_num_features)} / {len(num_features)}")
print(f"Kept: {selected_num_features}")

# New preprocessor with only the selected numeric features
preprocessor_fs = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler',  StandardScaler())
    ]), selected_num_features),
    ('cat', cat_transformer, cat_features)
])

fs_candidates = {
    'RF + Feature Selection':  RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1),
    'ET + Feature Selection':  ExtraTreesRegressor(n_estimators=400, random_state=42, n_jobs=-1),
}

X_train_fs = train[selected_num_features + cat_features]
X_test_fs  = test[selected_num_features + cat_features]

for name, reg in fs_candidates.items():
    pipe = Pipeline([('preprocessor', preprocessor_fs), ('regressor', reg)])
    scores = cross_val_score(pipe, X_train_fs, y_train, cv=5, scoring='r2', n_jobs=-1)
    results[name] = scores
    print(f"{name:35s}  mean R²={scores.mean():.4f}  std={scores.std():.4f}")
```

## Weighted Voting Ensemble & Final Predictions

```python
# Pick the best preprocessor+features combo found so far
all_v2_results = {k: results[k].mean() for k in ['RF + KNN Imputer', 'ET + KNN Imputer',
                                                   'RF + Feature Selection', 'ET + Feature Selection']}
best_v2_name = max(all_v2_results, key=all_v2_results.get)
print("CV R² summary:")
for k, v in sorted(all_v2_results.items(), key=lambda x: -x[1]):
    print(f"  {k:35s}  {v:.4f}")
print(f"\nBest: {best_v2_name} → {all_v2_results[best_v2_name]:.4f}")

# Weighted Voting: weight each estimator by its CV R²
w_rf_knn = results['RF + KNN Imputer'].mean()
w_et_knn = results['ET + KNN Imputer'].mean()
w_rf_fs  = results['RF + Feature Selection'].mean()
w_et_fs  = results['ET + Feature Selection'].mean()
w_base   = results.get(best_model_name_v2, np.array([0.0])).mean()  # best from previous round

pipe_rf_knn = Pipeline([('preprocessor', preprocessor_knn), ('regressor', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))])
pipe_et_knn = Pipeline([('preprocessor', preprocessor_knn), ('regressor', ExtraTreesRegressor(n_estimators=400, random_state=42, n_jobs=-1))])
pipe_rf_fs  = Pipeline([('preprocessor', preprocessor_fs),  ('regressor', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))])
pipe_et_fs  = Pipeline([('preprocessor', preprocessor_fs),  ('regressor', ExtraTreesRegressor(n_estimators=400, random_state=42, n_jobs=-1))])

weighted_voting = VotingRegressor([
    ('rf_knn', pipe_rf_knn),
    ('et_knn', pipe_et_knn),
    ('rf_fs',  pipe_rf_fs),
    ('et_fs',  pipe_et_fs),
], weights=[w_rf_knn, w_et_knn, w_rf_fs, w_et_fs])
# n_jobs removed from VotingRegressor: on Windows, nested parallelism (cross_val_score n_jobs=-1
# + VotingRegressor n_jobs=-1) causes workers to receive numpy arrays instead of DataFrames,
# breaking ColumnTransformer column-name lookup. Each RF/ET already uses n_jobs=-1 internally.

# Note: pipe_rf_fs and pipe_et_fs use X_train_fs, so we need a unified X_train
# Use full feature set and let each pipeline select its own columns internally
# Rebuild fs pipes to work on full X_train by including feature selection in the pipeline
from sklearn.feature_selection import SelectFromModel as SFM

preprocessor_fs_full = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler',  StandardScaler()),
    ]), num_features),
    ('cat', cat_transformer, cat_features)
], remainder='drop')

# VotingRegressor passes the same X to every sub-estimator, so we must use the
# full X_train. Each ColumnTransformer inside each sub-pipeline selects only the
# columns it needs (preprocessor_knn → num_features, preprocessor_fs → selected_num_features).
scores_wv = cross_val_score(weighted_voting, X_train, y_train, cv=5, scoring='r2', n_jobs=-1)
print(f"\nWeighted Voting  mean R²={scores_wv.mean():.4f}  std={scores_wv.std():.4f}")

# Train and predict with best model or weighted voting
if scores_wv.mean() > all_v2_results[best_v2_name]:
    print("\n→ Using Weighted Voting for final predictions")
    weighted_voting.fit(X_train, y_train)
    preds_v2 = weighted_voting.predict(X_test)
else:
    print(f"\n→ Using {best_v2_name} for final predictions")
    reg_cls = RandomForestRegressor if 'RF' in best_v2_name else ExtraTreesRegressor
    prep    = preprocessor_knn if 'KNN' in best_v2_name else preprocessor_fs
    X_tr    = X_train if 'KNN' in best_v2_name else X_train_fs
    X_te    = X_test  if 'KNN' in best_v2_name else X_test_fs
    best_pipe_final = Pipeline([('preprocessor', prep), ('regressor', reg_cls(n_estimators=400, random_state=42, n_jobs=-1))])
    best_pipe_final.fit(X_tr, y_train)
    preds_v2 = best_pipe_final.predict(X_te)

output_v2 = pd.DataFrame({'ID': test['ID'], 'target': preds_v2})
output_v2_path = '../submissions/predictions_v2.csv'
output_v2.to_csv(output_v2_path, index=False)
print(f"\nPredictions saved to: {output_v2_path}")
output_v2.head(10)
```

# Improving 3 - Detección de Outliers (GMM) + PCA

Aplicamos técnicas de **aprendizaje no supervisado** sobre el pipeline supervisado, siguiendo la metodología:

- **Outlier Detection (GMM)**: Entrenamos un GMM sobre el espacio de features numéricas preprocesadas. Usamos `score_samples()` para calcular la log-probabilidad de cada muestra y eliminamos el bottom 5% (muestras anómalas), tal como se hizo en el Ejercicio 3 del homework.
- **PCA**: Añadimos PCA como paso de reducción de dimensionalidad dentro del pipeline sklearn, determinando el número de componentes que explican el 95% de la varianza (Ejercicio 2 del homework). PCA puede reducir ruido y correlaciones, aunque para RF el beneficio varía según el dataset.

```python
from sklearn.mixture import GaussianMixture
from sklearn.base import clone
import warnings
warnings.filterwarnings('ignore')

# ── 1. Preprocesado de X_train para el GMM (solo numéricas) ──────────────────
#    Usamos KNNImputer + StandardScaler — mismo enfoque que el homework pero
#    solo sobre las features numéricas para que el GMM trabaje en escala uniforme
preproc_for_gmm = Pipeline([
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler',  StandardScaler())
])
X_train_gmm = preproc_for_gmm.fit_transform(X_train[num_features])
print(f"Shape para GMM: {X_train_gmm.shape}")

# ── 2. Selección del nº de componentes con BIC (igual que Ejercicio 3 homework) ─
bic_vals_gmm = []
n_range_gmm  = range(2, 12)
for n in n_range_gmm:
    gmm_tmp = GaussianMixture(n_components=n, n_init=10, random_state=42, reg_covar=1e-3)
    gmm_tmp.fit(X_train_gmm)
    bic_vals_gmm.append(gmm_tmp.bic(X_train_gmm))

best_n_gmm = list(n_range_gmm)[np.argmin(bic_vals_gmm)]
print(f"Nº óptimo de componentes GMM (BIC): {best_n_gmm}")

plt.figure(figsize=(8, 4))
plt.plot(list(n_range_gmm), bic_vals_gmm, 'b-o')
plt.axvline(best_n_gmm, color='r', linestyle='--', label=f'Óptimo = {best_n_gmm}')
plt.xlabel('Nº componentes GMM'); plt.ylabel('BIC')
plt.title('GMM Outlier Detection: selección por BIC')
plt.legend(); plt.tight_layout(); plt.show()

# ── 3. Ajustar el GMM y calcular log-probabilidades por muestra ──────────────
gmm_outlier = GaussianMixture(n_components=best_n_gmm, n_init=20,
                               random_state=42, reg_covar=1e-3)
gmm_outlier.fit(X_train_gmm)
log_probs_train = gmm_outlier.score_samples(X_train_gmm)

# ── 4. Eliminar el bottom 5% (metodología Ejercicio 3 del homework) ───────────
outlier_pct   = 5
threshold_gmm = np.percentile(log_probs_train, outlier_pct)
keep_mask     = log_probs_train >= threshold_gmm

X_train_clean = X_train[keep_mask].reset_index(drop=True)
y_train_clean = y_train[keep_mask].reset_index(drop=True)
print(f"\nOutliers eliminados : {(~keep_mask).sum()} ({(~keep_mask).mean()*100:.1f}%)")
print(f"Tamaño train limpio : {X_train_clean.shape[0]} (de {X_train.shape[0]})")

# ── 5. Distribución de log-probabilidades ────────────────────────────────────
plt.figure(figsize=(8, 4))
plt.hist(log_probs_train, bins=40, color='steelblue', edgecolor='white', alpha=0.8)
plt.axvline(threshold_gmm, color='r', linestyle='--',
            label=f'Percentil {outlier_pct} = {threshold_gmm:.2f}')
plt.xlabel('Log-probabilidad (GMM)'); plt.ylabel('Frecuencia')
plt.title('Log-probabilidad de las muestras de entrenamiento bajo el GMM')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
from sklearn.decomposition import PCA

# ── 1. Analizar varianza explicada por PCA sobre el espacio preprocesado ──────
#    Clonamos el preprocessor original para no alterar su estado
preprocessor_tmp = clone(preprocessor)
X_train_proc_all = preprocessor_tmp.fit_transform(X_train)   # (n_samples, n_features_total)

pca_var = PCA(random_state=42)
pca_var.fit(X_train_proc_all)
cumvar      = np.cumsum(pca_var.explained_variance_ratio_)
n_total_f   = X_train_proc_all.shape[1]
n_pca_95    = int(np.searchsorted(cumvar, 0.95)) + 1
n_pca_80    = int(np.searchsorted(cumvar, 0.80)) + 1

print(f"Dimensiones tras preprocesado : {n_total_f}")
print(f"Componentes PCA para 80% var  : {n_pca_80}")
print(f"Componentes PCA para 95% var  : {n_pca_95}")

plt.figure(figsize=(8, 4))
plt.plot(np.arange(1, n_total_f + 1), cumvar * 100, 'b-o', markersize=4)
plt.axhline(95, color='r', linestyle='--', label='95% varianza')
plt.axhline(80, color='g', linestyle='--', label='80% varianza')
plt.axvline(n_pca_95, color='r', linestyle=':', alpha=0.7)
plt.axvline(n_pca_80, color='g', linestyle=':', alpha=0.7)
plt.xlabel('Nº componentes PCA'); plt.ylabel('Varianza acumulada (%)')
plt.title('PCA sobre features preprocesadas: Varianza Explicada')
plt.legend(); plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()

# ── 2. Comparar 4 configuraciones (5-fold CV) ─────────────────────────────────
#    - baseline      : sin cambios
#    - outlier only  : solo se elimina el 5% de outliers (GMM)
#    - PCA only      : solo PCA (95% var) sobre datos completos
#    - outlier + PCA : combinación de ambas técnicas
rf_kw = dict(n_estimators=400, random_state=42, n_jobs=-1)

configs_v3 = {
    'RF – baseline (full data)': (
        Pipeline([('pre', preprocessor),
                  ('reg', RandomForestRegressor(**rf_kw))]),
        X_train, y_train
    ),
    'RF – outlier removal only': (
        Pipeline([('pre', preprocessor),
                  ('reg', RandomForestRegressor(**rf_kw))]),
        X_train_clean, y_train_clean
    ),
    f'RF – PCA({n_pca_95} comps) full data': (
        Pipeline([('pre', preprocessor),
                  ('pca', PCA(n_components=n_pca_95, random_state=42)),
                  ('reg', RandomForestRegressor(**rf_kw))]),
        X_train, y_train
    ),
    f'RF – outlier removal + PCA({n_pca_95})': (
        Pipeline([('pre', preprocessor),
                  ('pca', PCA(n_components=n_pca_95, random_state=42)),
                  ('reg', RandomForestRegressor(**rf_kw))]),
        X_train_clean, y_train_clean
    ),
}

results_v3 = {}
print("\nComparación CV R² (5-fold):")
print("─" * 65)
for name, (pipe, X_tr, y_tr) in configs_v3.items():
    scores = cross_val_score(pipe, X_tr, y_tr, cv=5, scoring='r2', n_jobs=-1)
    results_v3[name] = scores.mean()
    print(f"  {name:48s}  {scores.mean():.4f} ± {scores.std():.4f}")
```

```python
# ── Mejor configuración v3 → entrenar sobre todo el conjunto y predecir ───────
best_v3_name = max(results_v3, key=results_v3.get)
print(f"Mejor config v3 : '{best_v3_name}'")
print(f"CV R²           : {results_v3[best_v3_name]:.4f}")

best_v3_pipe, X_tr_v3, y_tr_v3 = configs_v3[best_v3_name]
best_v3_pipe.fit(X_tr_v3, y_tr_v3)
preds_v3 = best_v3_pipe.predict(X_test)

output_v3 = pd.DataFrame({'ID': test['ID'], 'target': preds_v3})
output_v3_path = '../submissions/predictions_v3.csv'
output_v3.to_csv(output_v3_path, index=False)
print(f"\nPredicciones guardadas en: {output_v3_path}")
output_v3.head(10)
```

# Improving 4 - Cluster Features (GMM→supervisado) + Stacking

Dos técnicas complementarias para superar el 0.856:

1. **Cluster features (GMM → supervisado)**: Usamos el GMM ya entrenado en Improving 3 para calcular `predict_proba` → probabilidades de pertenencia a cada cluster. Estas se añaden como nuevas features al dataset, transfiriendo directamente el conocimiento no supervisado al modelo supervisado.
2. **Stacking Ensemble**: Meta-learner Ridge entrenado sobre las predicciones out-of-fold de RF y ET como base learners. Más potente que el weighted voting porque el meta-learner aprende a combinar los modelos de forma óptima en lugar de usar pesos fijos.

```python
import warnings
warnings.filterwarnings('ignore')

# ── 1. Cluster features: GMM predict_proba → nuevas features ─────────────────
#    El GMM de Improving 3 aprendió la estructura no supervisada del espacio
#    de features. Las probabilidades de pertenencia a cada cluster son features
#    informativas que el modelo supervisado puede aprovechar.

# Preprocesar X_test con el mismo preproc_for_gmm (ya ajustado sobre X_train)
X_test_gmm   = preproc_for_gmm.transform(X_test[num_features])

gmm_probs_train = gmm_outlier.predict_proba(X_train_gmm)   # (n_train, K_gmm)
gmm_probs_test  = gmm_outlier.predict_proba(X_test_gmm)    # (n_test,  K_gmm)

gmm_cols = [f'gmm_p{i}' for i in range(best_n_gmm)]

X_train_aug = X_train.copy()
X_test_aug  = X_test.copy()
for j, col_name in enumerate(gmm_cols):
    X_train_aug[col_name] = gmm_probs_train[:, j]
    X_test_aug[col_name]  = gmm_probs_test[:, j]

print(f"Features originales : {X_train.shape[1]}")
print(f"Features augmentadas: {X_train_aug.shape[1]}  (+{best_n_gmm} GMM cluster probs)")

# Preprocessor ampliado que incluye las nuevas columnas GMM (ya son numéricas escaladas)
aug_num_features = num_features + gmm_cols   # todas las numéricas + cluster probs

preprocessor_aug = ColumnTransformer([
    ('normal', normal_transformer, [f for f in normal_features]),
    ('skewed', skewed_transformer, skewed_features),
    ('gmm',    StandardScaler(),   gmm_cols),          # cluster probs ya son [0,1]
    ('cat',    cat_transformer,    cat_features)
])

# CV baseline comparando con/sin cluster features
rf_aug = RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1)
for tag, prep, X_tr in [
        ('RF  sin cluster features', preprocessor,     X_train),
        ('RF  con cluster features', preprocessor_aug, X_train_aug)]:
    pipe = Pipeline([('pre', prep), ('reg', rf_aug)])
    sc = cross_val_score(pipe, X_tr, y_train, cv=5, scoring='r2', n_jobs=-1)
    print(f"{tag:40s}  R²={sc.mean():.4f} ± {sc.std():.4f}")
```

```python
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import Ridge

# ── Stacking: RF + ET como base learners → Ridge como meta-learner ────────────
#    En vez de promediar predicciones (voting), el meta-learner Ridge aprende
#    cómo combinar óptimamente las predicciones de RF y ET.
#    StackingRegressor usa cross_val_predict internamente (cv=5) para evitar
#    data leakage al construir el conjunto de entrenamiento del meta-learner.

pipe_rf_stack = Pipeline([('pre', preprocessor),     ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))])
pipe_et_stack = Pipeline([('pre', preprocessor),     ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))])
pipe_rf_aug   = Pipeline([('pre', preprocessor_aug), ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))])
pipe_et_aug   = Pipeline([('pre', preprocessor_aug), ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))])

stacking_base = StackingRegressor(
    estimators=[('rf', pipe_rf_stack), ('et', pipe_et_stack)],
    final_estimator=Ridge(alpha=1.0),
    cv=5,
    n_jobs=1      # evitar nested parallelism en Windows
)
stacking_aug = StackingRegressor(
    estimators=[('rf', pipe_rf_aug), ('et', pipe_et_aug)],
    final_estimator=Ridge(alpha=1.0),
    cv=5,
    n_jobs=1
)

# ── Comparativa ───────────────────────────────────────────────────────────────
results_v4 = {}
configs_v4 = {
    'RF  – baseline':               (Pipeline([('pre', preprocessor),     ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))]),  X_train,     y_train),
    'ET  – baseline':               (Pipeline([('pre', preprocessor),     ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))]),   X_train,     y_train),
    'RF  – cluster features GMM':   (pipe_rf_aug,                                                                                                           X_train_aug, y_train),
    'ET  – cluster features GMM':   (pipe_et_aug,                                                                                                           X_train_aug, y_train),
    'Stacking RF+ET (sin cluster)': (stacking_base,                                                                                                         X_train,     y_train),
    'Stacking RF+ET (con cluster)': (stacking_aug,                                                                                                          X_train_aug, y_train),
}

for name, (model, X_tr, y_tr) in configs_v4.items():
    sc = cross_val_score(model, X_tr, y_tr, cv=5, scoring='r2', n_jobs=-1)
    results_v4[name] = (sc, X_tr)
    print(f"{name:40s}  R²={sc.mean():.4f} ± {sc.std():.4f}")

print()
best_v4_name = max(results_v4, key=lambda k: results_v4[k][0].mean())
best_v4_sc, best_v4_X = results_v4[best_v4_name]
print(f"Mejor modelo v4: {best_v4_name}  R²={best_v4_sc.mean():.4f}")
```

```python
# ── Predicciones v4: entrenar el mejor modelo sobre todo el train ─────────────
best_v4_config = {
    'RF  – baseline':               (Pipeline([('pre', preprocessor),     ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))]),  X_train,     X_test),
    'ET  – baseline':               (Pipeline([('pre', preprocessor),     ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))]),   X_train,     X_test),
    'RF  – cluster features GMM':   (Pipeline([('pre', preprocessor_aug), ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))]),  X_train_aug, X_test_aug),
    'ET  – cluster features GMM':   (Pipeline([('pre', preprocessor_aug), ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))]),   X_train_aug, X_test_aug),
    'Stacking RF+ET (sin cluster)': (StackingRegressor(estimators=[('rf', pipe_rf_stack), ('et', pipe_et_stack)], final_estimator=Ridge(alpha=1.0), cv=5, n_jobs=1), X_train,     X_test),
    'Stacking RF+ET (con cluster)': (StackingRegressor(estimators=[('rf', pipe_rf_aug),   ('et', pipe_et_aug)],   final_estimator=Ridge(alpha=1.0), cv=5, n_jobs=1), X_train_aug, X_test_aug),
}

model_v4, X_tr_v4, X_te_v4 = best_v4_config[best_v4_name]

final_model_v4 = TransformedTargetRegressor(
    regressor=model_v4,
    transformer=PowerTransformer(method='yeo-johnson')
)
final_model_v4.fit(X_tr_v4, y_train)
preds_v4 = final_model_v4.predict(X_te_v4)

output_path_v4 = '../submissions/predictions_v4.csv'
pd.DataFrame({'ID': test['ID'], 'target': preds_v4}).to_csv(output_path_v4, index=False)
print(f"predictions_v4.csv guardado  ({best_v4_name})")
print(f"  R² CV: {best_v4_sc.mean():.4f} ± {best_v4_sc.std():.4f}")
```

# Improving 5 - SVR + IterativeImputer + Stacking diverso

Estrategias de mejora sin boosting ni feature engineering:

1. **IterativeImputer (MICE)**: Imputa cada feature como función del resto mediante regresión iterativa. Más preciso que KNN cuando las features están correlacionadas entre sí, porque modela la relación multivariante completa.
2. **SVR (Support Vector Regression)**: Algoritmo completamente distinto a RF/ET — busca un hiperplano de margen máximo en el espacio de features. Aporta diversidad real al ensemble porque sus errores no están correlacionados con los de los árboles.
3. **Stacking RF + ET + SVR → Ridge**: Añadir SVR como tercer base learner enriquece el meta-learner con una fuente de predicciones ortogonal a los dos árboles.
4. **Optimización del umbral GMM**: El percentil 5% para la detección de outliers fue fijo. Probamos varios umbrales (1%, 3%, 5%, 8%, 10%) para encontrar el que maximiza R² en CV.

```python
from sklearn.experimental import enable_iterative_imputer   # noqa — activa IterativeImputer
from sklearn.impute import IterativeImputer
from sklearn.svm import SVR
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import Ridge
import warnings
warnings.filterwarnings('ignore')

# ── 1. IterativeImputer (MICE) ────────────────────────────────────────────────
#    Imputa cada feature como regresión sobre el resto, iterando hasta
#    convergencia. Captura correlaciones entre features que KNN ignora.
iterative_imputer = IterativeImputer(max_iter=10, random_state=42)

normal_transformer_mice = Pipeline([
    ('imputer', IterativeImputer(max_iter=10, random_state=42)),
    ('scaler',  StandardScaler())
])
skewed_transformer_mice = Pipeline([
    ('imputer', IterativeImputer(max_iter=10, random_state=42)),
    ('power',   PowerTransformer(method='yeo-johnson', standardize=True))
])

preprocessor_mice = ColumnTransformer([
    ('normal', normal_transformer_mice, normal_features),
    ('skewed', skewed_transformer_mice, skewed_features),
    ('cat',    cat_transformer,         cat_features)
])

# ── 2. SVR como base learner ──────────────────────────────────────────────────
#    SVR necesita features escaladas (ya las proporciona el preprocessor).
#    Usamos kernel RBF — captura relaciones no lineales sin árboles.
pipe_svr = Pipeline([
    ('pre', preprocessor),
    ('reg', SVR(kernel='rbf', C=10.0, epsilon=0.1, gamma='scale'))
])
pipe_svr_mice = Pipeline([
    ('pre', preprocessor_mice),
    ('reg', SVR(kernel='rbf', C=10.0, epsilon=0.1, gamma='scale'))
])

# ── 3. Comparativa ────────────────────────────────────────────────────────────
results_v5 = {}
configs_v5 = {
    'RF  – IterativeImputer (MICE)':        (Pipeline([('pre', preprocessor_mice), ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))]), X_train, y_train),
    'ET  – IterativeImputer (MICE)':        (Pipeline([('pre', preprocessor_mice), ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))]), X_train, y_train),
    'SVR – preprocessor estándar':          (pipe_svr,                                                                                                             X_train, y_train),
    'SVR – IterativeImputer (MICE)':        (pipe_svr_mice,                                                                                                        X_train, y_train),
}

for name, (model, X_tr, y_tr) in configs_v5.items():
    sc = cross_val_score(model, X_tr, y_tr, cv=5, scoring='r2', n_jobs=-1)
    results_v5[name] = sc
    print(f"{name:45s}  R²={sc.mean():.4f} ± {sc.std():.4f}")
```

```python
# ── 4. Stacking RF + ET + SVR → Ridge ────────────────────────────────────────
#    RF y ET cometen errores similares (ambos son árboles de decisión).
#    SVR con kernel RBF tiene un patrón de error diferente → el meta-learner
#    puede aprender a corregir los puntos donde árboles fallan más.

# Usamos el mejor preprocessor encontrado para RF/ET (KNN o MICE según resultados_v5)
best_mice_rf = results_v5['RF  – IterativeImputer (MICE)'].mean()
best_knn_rf  = results_v4.get('RF  – baseline', (np.array([0]),))[0].mean()
prep_best = preprocessor_mice if best_mice_rf > best_knn_rf else preprocessor

pipe_rf_s5  = Pipeline([('pre', prep_best), ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))])
pipe_et_s5  = Pipeline([('pre', prep_best), ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))])
pipe_svr_s5 = Pipeline([('pre', prep_best), ('reg', SVR(kernel='rbf', C=10.0, epsilon=0.1, gamma='scale'))])

stacking_3 = StackingRegressor(
    estimators=[('rf', pipe_rf_s5), ('et', pipe_et_s5), ('svr', pipe_svr_s5)],
    final_estimator=Ridge(alpha=1.0),
    cv=5,
    n_jobs=1   # evitar nested parallelism en Windows
)

sc_s3 = cross_val_score(stacking_3, X_train, y_train, cv=5, scoring='r2', n_jobs=-1)
results_v5['Stacking RF+ET+SVR → Ridge'] = sc_s3
print(f"{'Stacking RF+ET+SVR → Ridge':45s}  R²={sc_s3.mean():.4f} ± {sc_s3.std():.4f}")

# ── 5. Optimización del umbral de outliers GMM ────────────────────────────────
#    Probamos distintos percentiles de corte para el GMM y evaluamos cada uno
pct_candidates = [1, 3, 5, 8, 10]
print("\nOptimización umbral GMM (RF baseline):")
best_pct, best_pct_score = 5, -np.inf
for pct in pct_candidates:
    thr = np.percentile(log_probs_train, pct)
    mask = log_probs_train >= thr
    X_cl = X_train[mask].reset_index(drop=True)
    y_cl = y_train[mask].reset_index(drop=True)
    pipe_tmp = Pipeline([('pre', preprocessor), ('reg', RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1))])
    sc_pct = cross_val_score(pipe_tmp, X_cl, y_cl, cv=5, scoring='r2', n_jobs=-1)
    tag = ' ← actual' if pct == 5 else ''
    print(f"  percentil {pct:2d}%  (eliminados={( ~mask).sum():3d})  R²={sc_pct.mean():.4f} ± {sc_pct.std():.4f}{tag}")
    if sc_pct.mean() > best_pct_score:
        best_pct_score = sc_pct.mean()
        best_pct = pct

print(f"\nMejor umbral GMM: percentil {best_pct}%  R²={best_pct_score:.4f}")

# ── Resumen v5 ─────────────────────────────────────────────────────────────────
print("\n" + "─" * 60)
print("Resumen Improving 5:")
for k, v in sorted(results_v5.items(), key=lambda x: -x[1].mean()):
    print(f"  {k:45s}  {v.mean():.4f}")
best_v5_name = max(results_v5, key=lambda k: results_v5[k].mean())
print(f"\nMejor: {best_v5_name}  R²={results_v5[best_v5_name].mean():.4f}")
```

```python
# ── Predicciones v5: mejor modelo → predictions_v5.csv ───────────────────────
best_v5_model_map = {
    'RF  – IterativeImputer (MICE)':   Pipeline([('pre', preprocessor_mice), ('reg', RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1))]),
    'ET  – IterativeImputer (MICE)':   Pipeline([('pre', preprocessor_mice), ('reg', ExtraTreesRegressor( n_estimators=400, random_state=42, n_jobs=-1))]),
    'SVR – preprocessor estándar':     pipe_svr,
    'SVR – IterativeImputer (MICE)':   pipe_svr_mice,
    'Stacking RF+ET+SVR → Ridge':      StackingRegressor(
                                           estimators=[('rf', pipe_rf_s5), ('et', pipe_et_s5), ('svr', pipe_svr_s5)],
                                           final_estimator=Ridge(alpha=1.0), cv=5, n_jobs=1),
}

# Entrenar el mejor modelo con TransformedTargetRegressor
final_model_v5 = TransformedTargetRegressor(
    regressor=best_v5_model_map[best_v5_name],
    transformer=PowerTransformer(method='yeo-johnson')
)
final_model_v5.fit(X_train, y_train)
preds_v5 = final_model_v5.predict(X_test)

output_path_v5 = '../submissions/predictions_v5.csv'
pd.DataFrame({'ID': test['ID'], 'target': preds_v5}).to_csv(output_path_v5, index=False)
print(f"predictions_v5.csv guardado  ({best_v5_name})")
print(f"  R² CV: {results_v5[best_v5_name].mean():.4f} ± {results_v5[best_v5_name].std():.4f}")
```

## Sequential Feature Selection (Forward & Backward)

La selección basada en MI que usamos antes es un **método filtro**: puntúa cada feature de forma independiente sin considerar cómo interaccionan entre sí. Dos features pueden tener MI alta pero ser redundantes entre ellas (aportan la misma información).

**SequentialFeatureSelector** es un **método wrapper**: en cada paso ajusta y evalúa el modelo real con CV, así el subconjunto seleccionado está optimizado directamente para el R² del RF:

- **Forward**: empieza sin features, añade en cada paso la que más mejora el CV R²
- **Backward**: empieza con todas, elimina en cada paso la que menos perjudica al CV R²

> Se usa `n_estimators=150` para acelerar el proceso (cada paso requiere múltiples fits de CV).

```python
from sklearn.feature_selection import SequentialFeatureSelector
import warnings
warnings.filterwarnings('ignore')

# RF ligero para acelerar SFS (cada paso = múltiples CV fits)
rf_sfs = RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1)

# Preprocesador intermedio: imputa + escala todas las numéricas antes del SFS
# SFS opera sobre el array resultante; recuperamos los nombres al final
preprocessor_sfs = ColumnTransformer([
    ('num', Pipeline([('imp', KNNImputer(n_neighbors=5)), ('sc', StandardScaler())]), num_features),
    ('cat', cat_transformer, cat_features)
])
X_train_proc_sfs = preprocessor_sfs.fit_transform(X_train)
X_test_proc_sfs  = preprocessor_sfs.transform(X_test)

# Nombres de columna tras el preprocessor (num primero, luego OHE cats)
ohe_cols_sfs = (preprocessor_sfs
                .named_transformers_['cat']
                .named_steps['encoder']
                .get_feature_names_out(cat_features).tolist())
all_cols_sfs = num_features + ohe_cols_sfs

# ── Forward SFS ───────────────────────────────────────────────────────────────
print("Ejecutando Forward SFS  (puede tardar ~2-4 min)…")
sfs_forward = SequentialFeatureSelector(
    rf_sfs,
    direction='forward',
    n_features_to_select='auto',   # para cuando añadir más no mejora
    tol=1e-4,
    scoring='r2',
    cv=5,
    n_jobs=-1
)
sfs_forward.fit(X_train_proc_sfs, y_train)
forward_mask  = sfs_forward.get_support()
forward_feats = [c for c, keep in zip(all_cols_sfs, forward_mask) if keep and c in num_features]
print(f"Forward seleccionó {len(forward_feats)} features: {forward_feats}")

# ── Backward SFS ──────────────────────────────────────────────────────────────
print("\nEjecutando Backward SFS (puede tardar ~3-5 min)…")
sfs_backward = SequentialFeatureSelector(
    rf_sfs,
    direction='backward',
    n_features_to_select='auto',
    tol=1e-4,
    scoring='r2',
    cv=5,
    n_jobs=-1
)
sfs_backward.fit(X_train_proc_sfs, y_train)
backward_mask  = sfs_backward.get_support()
backward_feats = [c for c, keep in zip(all_cols_sfs, backward_mask) if keep and c in num_features]
print(f"Backward seleccionó {len(backward_feats)} features: {backward_feats}")

# ── Comparativa con los subconjuntos anteriores ───────────────────────────────
print("\nComparación CV R² (5-fold):")
print("─" * 60)
sfs_sets = {
    f'Forward  SFS  ({len(forward_feats)} feats)':  forward_feats,
    f'Backward SFS  ({len(backward_feats)} feats)': backward_feats,
    f'Filter MI     ({len(mi_selected)} feats)':    mi_selected,
    f'RF Importance ({len(rf_selected)} feats)':    rf_selected,
}
results_sfs = {}
for name, feats in sfs_sets.items():
    X_fs = X_train[feats + cat_features]
    pipe = make_pipe_fs(feats)
    sc = cross_val_score(pipe, X_fs, y_train, cv=5, scoring='r2', n_jobs=-1)
    results_sfs[name] = (sc, feats)
    print(f"  {name:35s}  R²={sc.mean():.4f} ± {sc.std():.4f}")

best_sfs_name = max(results_sfs, key=lambda k: results_sfs[k][0].mean())
best_sfs_sc, best_sfs_feats = results_sfs[best_sfs_name]
print(f"\nMejor método: '{best_sfs_name}'  R²={best_sfs_sc.mean():.4f}")
```

```python
# ── Tuning del mejor subconjunto SFS ─────────────────────────────────────────
X_train_sfs_best = X_train[best_sfs_feats + cat_features]
X_test_sfs_best  = X_test[ best_sfs_feats + cat_features]

preprocessor_sfs_best = ColumnTransformer([
    ('num', Pipeline([('imp', KNNImputer(n_neighbors=5)), ('sc', StandardScaler())]), best_sfs_feats),
    ('cat', cat_transformer, cat_features)
])
pipe_sfs_tune = Pipeline([
    ('pre', preprocessor_sfs_best),
    ('reg', RandomForestRegressor(random_state=42, n_jobs=-1))
])

param_grid_sfs = {
    'reg__n_estimators':      [400, 600, 800, 1000],
    'reg__max_depth':         [None, 15, 25, 35],
    'reg__min_samples_leaf':  [1, 2, 3, 5],
    'reg__min_samples_split': [2, 4, 8],
    'reg__max_features':      ['sqrt', 'log2', 0.4, 0.6],
    'reg__bootstrap':         [True, False],
}

search_sfs = RandomizedSearchCV(
    pipe_sfs_tune,
    param_distributions=param_grid_sfs,
    n_iter=80,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1,
    verbose=1
)
search_sfs.fit(X_train_sfs_best, y_train)
print(f"\nMejor R² CV (RF + {best_sfs_name}, tuned): {search_sfs.best_score_:.4f}")
print(f"Mejora sobre baseline 0.856: {search_sfs.best_score_ - 0.856:+.4f}")
print(f"Best params: {search_sfs.best_params_}")

# ── Predicciones finales ──────────────────────────────────────────────────────
preds_sfs = search_sfs.best_estimator_.predict(X_test_sfs_best)
output_path_sfs = '../submissions/predictions_v7.csv'
pd.DataFrame({'ID': test['ID'], 'target': preds_sfs}).to_csv(output_path_sfs, index=False)
print(f"\npredictions_v7.csv guardado  (RF + {best_sfs_name}, tuned)")
```

# Improving 7 — Grand Ensemble: SFS + GMM Features + Stacking + TTR consistente

Hasta ahora cada técnica se probó por separado. Aquí **combinamos todo lo que mejor funcionó** y corregimos una inconsistencia importante: en los Improving anteriores `TransformedTargetRegressor` (TTR) **no se aplicaba durante el CV**, solo al predecir. Eso significa que el modelo fue evaluado en condiciones distintas a las del entrenamiento final.

**Combinación completa:**
- Features: subconjunto SFS (wrapper, mejor que filter) + probabilidades GMM (conocimiento no supervisado)
- Ensemble: Stacking RF + ET + SVR → meta-learner (más potente que weighted voting)
- TTR aplicado **dentro** del CV: el modelo se evalúa exactamente igual que se entrena
- Meta-learner comparado: Ridge vs ElasticNet vs RF ligero

```python
from sklearn.linear_model import ElasticNet
import warnings
warnings.filterwarnings('ignore')

# ── Helper: envuelve cualquier pipeline en TTR (Yeo-Johnson) ──────────────────
def with_ttr(pipe):
    return TransformedTargetRegressor(
        regressor=pipe,
        transformer=PowerTransformer(method='yeo-johnson')
    )

# ── 1. Feature set: SFS numéricos + GMM cluster probs ────────────────────────
# best_sfs_feats puede incluir nombres OHE generados por SFS; nos quedamos
# solo con los que son features numéricas originales del dataset
combo_num = [f for f in best_sfs_feats if f in num_features]
combo_all = combo_num + gmm_cols + cat_features

X_train_c7 = X_train_aug[combo_all]
X_test_c7  = X_test_aug[ combo_all]

preprocessor_c7 = ColumnTransformer([
    ('num', Pipeline([('imp', KNNImputer(n_neighbors=5)), ('sc', StandardScaler())]), combo_num),
    ('gmm', StandardScaler(),                                                          gmm_cols),
    ('cat', cat_transformer,                                                           cat_features),
])

print(f"Feature set Improving 7:")
print(f"  SFS numéricas : {len(combo_num)} → {combo_num}")
print(f"  GMM cluster   : {len(gmm_cols)}")
print(f"  Categóricas   : {len(cat_features)}")
print(f"  TOTAL         : {len(combo_all)}")

# ── 2. Base learners ──────────────────────────────────────────────────────────
kw = dict(random_state=42, n_jobs=-1)
pipe_rf_c7  = Pipeline([('pre', preprocessor_c7), ('reg', RandomForestRegressor(n_estimators=500, **kw))])
pipe_et_c7  = Pipeline([('pre', preprocessor_c7), ('reg', ExtraTreesRegressor( n_estimators=500, **kw))])
pipe_svr_c7 = Pipeline([('pre', preprocessor_c7), ('reg', SVR(kernel='rbf', C=50.0, epsilon=0.1, gamma='scale'))])

# ── 3. Cuatro configuraciones para comparar ───────────────────────────────────
def make_stack(meta):
    return StackingRegressor(
        estimators=[('rf', pipe_rf_c7), ('et', pipe_et_c7), ('svr', pipe_svr_c7)],
        final_estimator=meta,
        cv=5,
        n_jobs=1   # evitar nested parallelism en Windows
    )

configs_v7 = {
    'RF  – SFS+GMM+TTR':                         with_ttr(pipe_rf_c7),
    'ET  – SFS+GMM+TTR':                         with_ttr(pipe_et_c7),
    'Stack RF+ET+SVR→Ridge    (SFS+GMM+TTR)':    with_ttr(make_stack(Ridge(alpha=1.0))),
    'Stack RF+ET+SVR→ElasticNet (SFS+GMM+TTR)':  with_ttr(make_stack(ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=2000))),
    'Stack RF+ET+SVR→RF_meta  (SFS+GMM+TTR)':    with_ttr(make_stack(RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42))),
}

# ── 4. 5-fold CV con TTR aplicado consistentemente ───────────────────────────
print("\nComparativa CV R² (5-fold) — TTR aplicado durante el CV:")
print("─" * 70)
cv_results_v7 = {}
for name, model in configs_v7.items():
    sc = cross_val_score(model, X_train_c7, y_train, cv=5, scoring='r2', n_jobs=-1)
    cv_results_v7[name] = sc
    print(f"  {name:50s}  R²={sc.mean():.4f} ± {sc.std():.4f}")

print()
best_v7_name = max(cv_results_v7, key=lambda k: cv_results_v7[k].mean())
best_v7_sc   = cv_results_v7[best_v7_name]
print(f"Mejor modelo v7 : '{best_v7_name}'")
print(f"R² CV           : {best_v7_sc.mean():.4f} ± {best_v7_sc.std():.4f}")
print(f"Δ vs 0.856      : {best_v7_sc.mean() - 0.856:+.4f}")
```

```python
# ── Tuning del mejor modelo v7 ────────────────────────────────────────────────
# Si el ganador es un modelo simple (RF/ET), se hace RandomizedSearchCV.
# Si es Stacking, se tunan los hiperparámetros de los base learners.

meta_map = {
    'Stack RF+ET+SVR→Ridge    (SFS+GMM+TTR)':   Ridge(alpha=1.0),
    'Stack RF+ET+SVR→ElasticNet (SFS+GMM+TTR)': ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=2000),
    'Stack RF+ET+SVR→RF_meta  (SFS+GMM+TTR)':   RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42),
}

if best_v7_name in meta_map:
    # El ganador es un modelo de stacking → tunar los base learners RF/ET y SVR
    param_dist_v7 = {
        'regressor__rf__reg__n_estimators':      [300, 500, 700],
        'regressor__rf__reg__min_samples_leaf':  [1, 2, 4],
        'regressor__rf__reg__max_features':      ['sqrt', 0.5, 0.7],
        'regressor__et__reg__n_estimators':      [300, 500, 700],
        'regressor__et__reg__min_samples_leaf':  [1, 2, 4],
        'regressor__svr__reg__C':                [10, 50, 100],
        'regressor__svr__reg__epsilon':          [0.05, 0.1, 0.2],
    }
    best_stack_tune = with_ttr(StackingRegressor(
        estimators=[('rf', pipe_rf_c7), ('et', pipe_et_c7), ('svr', pipe_svr_c7)],
        final_estimator=meta_map[best_v7_name],
        cv=5, n_jobs=1
    ))
    search_v7 = RandomizedSearchCV(
        best_stack_tune,
        param_distributions=param_dist_v7,
        n_iter=20, cv=5, scoring='r2',
        random_state=42, n_jobs=-1, verbose=1
    )
    search_v7.fit(X_train_c7, y_train)
    print(f"Tuned R²: {search_v7.best_score_:.4f}  (Δ vs 0.856: {search_v7.best_score_-0.856:+.4f})")
    print(f"Best params: {search_v7.best_params_}")
    model_v7_final = search_v7.best_estimator_
else:
    # El ganador es RF o ET individual → tunar directamente
    param_dist_v7 = {
        'regressor__reg__n_estimators':      [400, 600, 800, 1000],
        'regressor__reg__max_depth':         [None, 20, 30],
        'regressor__reg__min_samples_leaf':  [1, 2, 3],
        'regressor__reg__max_features':      ['sqrt', 'log2', 0.5],
        'regressor__reg__bootstrap':         [True, False],
    }
    reg_cls = RandomForestRegressor if 'RF' in best_v7_name else ExtraTreesRegressor
    base = with_ttr(Pipeline([('pre', preprocessor_c7), ('reg', reg_cls(random_state=42, n_jobs=-1))]))
    search_v7 = RandomizedSearchCV(
        base, param_distributions=param_dist_v7,
        n_iter=60, cv=5, scoring='r2',
        random_state=42, n_jobs=-1, verbose=1
    )
    search_v7.fit(X_train_c7, y_train)
    print(f"Tuned R²: {search_v7.best_score_:.4f}  (Δ vs 0.856: {search_v7.best_score_-0.856:+.4f})")
    print(f"Best params: {search_v7.best_params_}")
    model_v7_final = search_v7.best_estimator_

# ── Predicciones finales v8 ───────────────────────────────────────────────────
preds_v8 = model_v7_final.predict(X_test_c7)
output_path_v8 = '../submissions/predictions_v8.csv'
pd.DataFrame({'ID': test['ID'], 'target': preds_v8}).to_csv(output_path_v8, index=False)
print(f"\npredictions_v8.csv guardado  ({best_v7_name}, tuned)")
```

# Improving 8 — Best Feature Subset + Tuning Agresivo + Seed Averaging

El Improving 7 empeoró el score (0.84793 vs 0.85596) por tres razones:
1. Las features GMM añaden ruido para este dataset — los clusters no son discriminativos para el target
2. SVR con C=50 sin optimizar previa arrastra el stacking hacia abajo
3. El filtro `best_sfs_feats ∩ num_features` perdió features OHE que SFS había seleccionado

**Estrategia de recuperación y mejora:**
- Evaluar **todos** los subconjuntos (RF Importance, MI, Intersección, Unión, Forward SFS, Backward SFS) con RF y ET, con y sin TTR → encontrar el realmente mejor
- `RandomizedSearchCV` con **n_iter=150** incluyendo `ccp_alpha` (pruning de árboles, nunca probado)
- **Seed Averaging**: entrenar con 10 semillas distintas y promediar predicciones — reduce la varianza aleatoria de los árboles sin cambiar la arquitectura del modelo

```python
from sklearn.base import clone
import warnings
warnings.filterwarnings('ignore')

# ── Helper: pipeline con TTR opcional ────────────────────────────────────────
def make_eval_pipe(feats, reg_cls, n_est=400, use_ttr=True):
    prep = ColumnTransformer([
        ('num', Pipeline([('imp', KNNImputer(n_neighbors=5)), ('sc', StandardScaler())]), feats),
        ('cat', cat_transformer, cat_features)
    ])
    base = Pipeline([('pre', prep), ('reg', reg_cls(n_estimators=n_est, random_state=42, n_jobs=-1))])
    if use_ttr:
        return TransformedTargetRegressor(regressor=base,
                                          transformer=PowerTransformer(method='yeo-johnson'))
    return base

# ── 1. Todos los subconjuntos × {RF, ET} × {con TTR, sin TTR} ────────────────
all_fs = {
    'RF Importance':  rf_selected,
    'MI':             mi_selected,
    'Intersección':   inter_sel,
    'Unión':          union_sel,
    'Forward SFS':    forward_feats,
    'Backward SFS':   backward_feats,
    'Todas':          num_features,
}

print(f"{'Config':50s}  R²(mean)  std")
print("─" * 68)
results_v8 = {}
for fs_name, feats in all_fs.items():
    if not feats:
        continue
    X_fs = X_train[feats + cat_features]
    for reg_name, reg_cls in [('RF', RandomForestRegressor), ('ET', ExtraTreesRegressor)]:
        for ttr, ttr_label in [(True, '+TTR'), (False, '    ')]:
            model = make_eval_pipe(feats, reg_cls, n_est=400, use_ttr=ttr)
            sc = cross_val_score(model, X_fs, y_train, cv=5, scoring='r2', n_jobs=-1)
            key = f'{reg_name}{ttr_label} | {fs_name}'
            results_v8[key] = {'sc': sc, 'feats': feats, 'reg_cls': reg_cls,
                                'use_ttr': ttr, 'X_tr': X_fs,
                                'X_te': X_test[feats + cat_features]}
            print(f"  {key:48s}  {sc.mean():.4f}   {sc.std():.4f}")

best_v8_key = max(results_v8, key=lambda k: results_v8[k]['sc'].mean())
bv8 = results_v8[best_v8_key]
print(f"\nMejor: '{best_v8_key}'  R²={bv8['sc'].mean():.4f}")
print(f"Δ vs 0.856: {bv8['sc'].mean()-0.856:+.4f}  |  target 0.85984  Δ={bv8['sc'].mean()-0.85984:+.4f}")
```

```python
# ── 2. Tuning agresivo del mejor config (n_iter=150 + ccp_alpha) ─────────────
best_feats_v8 = bv8['feats']
best_reg_v8   = bv8['reg_cls']
best_ttr_v8   = bv8['use_ttr']
X_tr_v8       = bv8['X_tr']
X_te_v8       = bv8['X_te']

prep_v8 = ColumnTransformer([
    ('num', Pipeline([('imp', KNNImputer(n_neighbors=5)), ('sc', StandardScaler())]), best_feats_v8),
    ('cat', cat_transformer, cat_features)
])

if best_ttr_v8:
    pipe_to_tune = TransformedTargetRegressor(
        regressor=Pipeline([('pre', prep_v8), ('reg', best_reg_v8(random_state=42, n_jobs=-1))]),
        transformer=PowerTransformer(method='yeo-johnson')
    )
    param_prefix = 'regressor__reg__'
else:
    pipe_to_tune = Pipeline([('pre', prep_v8), ('reg', best_reg_v8(random_state=42, n_jobs=-1))])
    param_prefix = 'reg__'

param_grid_v8 = {
    f'{param_prefix}n_estimators':      [400, 600, 800, 1000, 1500],
    f'{param_prefix}max_depth':         [None, 20, 30, 40],
    f'{param_prefix}min_samples_leaf':  [1, 2, 3, 4],
    f'{param_prefix}min_samples_split': [2, 4, 6],
    f'{param_prefix}max_features':      ['sqrt', 'log2', 0.3, 0.4, 0.5, 0.6],
    f'{param_prefix}bootstrap':         [True, False],
    f'{param_prefix}ccp_alpha':         [0.0, 0.00005, 0.0001, 0.0005],  # pruning — nunca probado
}

search_v8 = RandomizedSearchCV(
    pipe_to_tune,
    param_distributions=param_grid_v8,
    n_iter=150, cv=5, scoring='r2',
    random_state=42, n_jobs=-1, verbose=1
)
search_v8.fit(X_tr_v8, y_train)
print(f"\nTuned R²: {search_v8.best_score_:.4f}  (Δ vs 0.856: {search_v8.best_score_-0.856:+.4f})")
print(f"Δ vs target 0.85984: {search_v8.best_score_-0.85984:+.4f}")
print(f"Best params: {search_v8.best_params_}")
best_model_v8 = search_v8.best_estimator_
```

```python
# ── 3. Seed Averaging: promedio de predicciones con 10 semillas distintas ─────
#    Cada RF/ET tiene varianza por el muestreo aleatorio de árboles y features.
#    Entrenar con semillas distintas y promediar cancela esa varianza
#    (mismo efecto que tener 10x más árboles, pero más eficiente en memoria).

seeds = [42, 0, 7, 13, 99, 123, 256, 512, 777, 999]
best_params = search_v8.best_params_

# Extraer parámetros del estimador ganador para aplicarlos a cada semilla
seed_preds = []
for seed in seeds:
    model_s = clone(best_model_v8)
    # Ajustar la semilla en el parámetro correcto según TTR o no
    seed_param = f'{param_prefix}random_state'
    model_s.set_params(**{seed_param: seed})
    model_s.fit(X_tr_v8, y_train)
    seed_preds.append(model_s.predict(X_te_v8))

preds_v9 = np.mean(seed_preds, axis=0)

# Evaluar también el promedio en CV con las distintas semillas
print("Seed averaging — CV R² individual por semilla:")
cv_seed_scores = []
for seed in seeds:
    model_s = clone(best_model_v8)
    model_s.set_params(**{seed_param: seed})
    sc_s = cross_val_score(model_s, X_tr_v8, y_train, cv=5, scoring='r2', n_jobs=-1)
    cv_seed_scores.append(sc_s.mean())
    print(f"  seed={seed:4d}  R²={sc_s.mean():.4f}")
print(f"\nMedia de las 10 semillas: {np.mean(cv_seed_scores):.4f}  std={np.std(cv_seed_scores):.4f}")

# ── 4. Guardar predictions_v9.csv ────────────────────────────────────────────
output_path_v9 = '../submissions/predictions_v9.csv'
pd.DataFrame({'ID': test['ID'], 'target': preds_v9}).to_csv(output_path_v9, index=False)
print(f"\npredictions_v9.csv guardado  ({best_v8_key}, seed_avg=10)")
print(f"  Mejor CV individual: {search_v8.best_score_:.4f}")
print(f"  CV promedio 10 seeds: {np.mean(cv_seed_scores):.4f}")
```
